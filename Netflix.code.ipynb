```python


# NETFLIX VIEWING ANALYTICS PIPELINE
# ETL USING PYTHON OBJECT-ORIENTED PROGRAMMING



# 1. PROJECT OBJECTIVES


"""
OBJECTIVES:

1. Create Movie, User and WatchHistory classes.
2. Store Netflix viewing data using Python objects.
3. Extract data from watch history.
4. Transform data using Pandas.
5. Calculate total viewing events.
6. Calculate average movie ratings.
7. Identify the most watched movie.
8. Identify the most active user.
9. Generate an analytics report.
10. Visualise the results.
"""



# 2. IMPORT LIBRARIES

import pandas as pd
import matplotlib.pyplot as plt
import logging

from collections import Counter

print("All libraries imported successfully!")



# 3. CONFIGURE LOGGING


logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

logging.info("Netflix Analytics Pipeline Initialized")



# 4. DATA STORAGE - CREATE CLASSES



# Movie Class

class Movie:

    def __init__(self, movie_id, title, genre):

        self.movie_id = movie_id
        self.title = title
        self.genre = genre


# User Class

class User:

    def __init__(self, user_id, name):

        self.user_id = user_id
        self.name = name


# WatchHistory Class

class WatchHistory:

    def __init__(self, user, movie, rating):

        self.user = user
        self.movie = movie
        self.rating = rating


logging.info("All classes created successfully!")



# 5. CREATE MOVIE OBJECTS


movies = [

    Movie(1, "Inception", "Sci-Fi"),

    Movie(2, "Interstellar", "Sci-Fi"),

    Movie(3, "Titanic", "Romance")

]

logging.info("Movie objects created: %d", len(movies))



# 6. CREATE USER OBJECTS


users = [

    User(101, "Nisarg"),

    User(102, "Aryan"),

    User(103, "Dhruv")

]

logging.info("User objects created: %d", len(users))



# 7. CREATE WATCH HISTORY

history = [

    WatchHistory(users[0], movies[0], 5),

    WatchHistory(users[1], movies[0], 4),

    WatchHistory(users[2], movies[1], 5),

    WatchHistory(users[0], movies[1], 4),

    WatchHistory(users[1], movies[2], 5),

    WatchHistory(users[2], movies[0], 3)

]

logging.info("Watch history records created: %d", len(history))



# 8. ETL PIPELINE - EXTRACT


def extract_data(history):

    extracted_data = []

    for record in history:

        extracted_data.append({

            "User_ID": record.user.user_id,

            "User": record.user.name,

            "Movie_ID": record.movie.movie_id,

            "Movie": record.movie.title,

            "Genre": record.movie.genre,

            "Rating": record.rating

        })

    logging.info("Data extraction completed!")

    return extracted_data


# Execute extraction

raw_data = extract_data(history)



# 9. ETL PIPELINE - TRANSFORM


def transform_data(raw_data):

    # Convert extracted data into DataFrame

    df = pd.DataFrame(raw_data)

    # Total viewing events

    total_views = len(df)

    # Average movie ratings

    average_ratings = (

        df.groupby("Movie")["Rating"]

        .mean()

        .round(2)

    )

    # Count movie views

    movie_views = df["Movie"].value_counts()

    # Most watched movie

    most_watched_movie = movie_views.idxmax()

    # Count user activity

    user_activity = df["User"].value_counts()

    # Most active user

    most_active_user = user_activity.idxmax()

    # Genre-wise analysis

    genre_analysis = df.groupby("Genre").agg(

        Total_Views=("Movie", "count"),

        Average_Rating=("Rating", "mean")

    ).round(2)

    # Store all analytics

    analytics = {

        "dataframe": df,

        "total_views": total_views,

        "average_ratings": average_ratings,

        "movie_views": movie_views,

        "most_watched_movie": most_watched_movie,

        "user_activity": user_activity,

        "most_active_user": most_active_user,

        "genre_analysis": genre_analysis

    }

    logging.info("Data transformation completed!")

    return analytics


# Execute transformation

analytics = transform_data(raw_data)



# 10. ETL PIPELINE - LOAD / GENERATE REPORT


def load_report(analytics):

    print("\n")

    print("=" * 55)

    print("          NETFLIX DAILY ANALYTICS REPORT")

    print("=" * 55)

    # Total viewing activity

    print("\n1. TOTAL VIEWING EVENTS")

    print(analytics["total_views"])

    # Most watched movie

    print("\n2. MOST WATCHED MOVIE")

    print(analytics["most_watched_movie"])

    # Movie viewing statistics

    print("\n3. MOVIE VIEW COUNTS")

    print(analytics["movie_views"])

    # Most active user

    print("\n4. MOST ACTIVE USER")

    print(analytics["most_active_user"])

    # User activity

    print("\n5. USER ACTIVITY")

    print(analytics["user_activity"])

    # Average ratings

    print("\n6. AVERAGE MOVIE RATINGS")

    print(analytics["average_ratings"])

    # Genre analysis

    print("\n7. GENRE-WISE ANALYSIS")

    print(analytics["genre_analysis"])

    print("\n" + "=" * 55)

    print("             REPORT COMPLETED")

    print("=" * 55)


# Execute report

load_report(analytics)



# 11. COMPLETE ETL PIPELINE


def run_pipeline(history):

    logging.info("Starting complete ETL pipeline")

    # EXTRACT

    raw_data = extract_data(history)

    # TRANSFORM

    analytics = transform_data(raw_data)

    # LOAD

    load_report(analytics)

    logging.info("ETL pipeline executed successfully!")

    return analytics


# Run complete pipeline

final_analytics = run_pipeline(history)



# 12. DATA ANALYSIS


df = final_analytics["dataframe"]


# Display complete dataset

print("\nCOMPLETE NETFLIX DATASET")

display(df)



# 13. MOST WATCHED MOVIE ANALYSIS


print("\nMOST WATCHED MOVIE")

print(final_analytics["most_watched_movie"])

display(

    final_analytics["movie_views"]

    .to_frame("Total Views")

)



# 14. MOST ACTIVE USER ANALYSIS


print("\nMOST ACTIVE USER")

print(final_analytics["most_active_user"])

display(

    final_analytics["user_activity"]

    .to_frame("Total Watches")

)



# 15. AVERAGE MOVIE RATINGS


print("\nAVERAGE MOVIE RATINGS")

display(

    final_analytics["average_ratings"]

    .to_frame("Average Rating")

)



# 16. GENRE-WISE ANALYSIS


print("\nGENRE-WISE ANALYSIS")

display(final_analytics["genre_analysis"])



# 17. DATA VISUALIZATION - MOVIE POPULARITY


plt.figure(figsize=(9, 5))

final_analytics["movie_views"].plot(

    kind="bar",

    title="Netflix Movie Popularity",

    xlabel="Movie Name",

    ylabel="Total Views"

)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()



# 18. DATA VISUALIZATION - USER ACTIVITY


plt.figure(figsize=(9, 5))

final_analytics["user_activity"].plot(

    kind="bar",

    title="Netflix User Activity",

    xlabel="User Name",

    ylabel="Total Watches"

)

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()


# 19. DATA VISUALIZATION - GENRE ANALYSIS

plt.figure(figsize=(8, 5))

final_analytics["genre_analysis"]["Total_Views"].plot(

    kind="pie",

    autopct="%1.1f%%",

    title="Genre-wise Viewing Distribution"

)

plt.ylabel("")

plt.tight_layout()

plt.show()



# 20. SAVE TRANSFORMED DATA

df.to_csv("netflix_viewing_analytics.csv", index=False)

logging.info("Dataset successfully saved as CSV")



# 21. FINAL CONCLUSION


print("\n")

print("=" * 55)

print("                 PROJECT CONCLUSION")

print("=" * 55)

print("""

1. Successfully implemented Python OOP classes.

2. Stored movie, user and viewing history data.

3. Extracted structured data using ETL.

4. Transformed data using Pandas.

5. Calculated total viewing events and average ratings.

6. Identified the most watched movie.

7. Identified the most active user.

8. Performed genre-wise analysis.

9. Generated visualizations and analytics reports.

10. Exported the transformed dataset into CSV format.


FINAL RESULT:

Netflix Viewing Analytics Pipeline
EXECUTED SUCCESSFULLY!

""")

print("=" * 55)

```
